## Parte 4. Variables predictoras.

x --> Aquellas variables que nos ayudan a predecir
y --> Variavle objetivo

```python
dfTitanic.head(1)
```

```python
columnas= [
    "Pclass",
    "Sex",
    "Age",
    "Fare",
    "Embarked"
]
```

```python
X = dfTitanic[columnas]
Y = dfTitanic["Survived"]
```

## Parte 5. Limpieza de los datos

Realizaremos la imputacion de los datos, donde existan valores nulos.

```python
X.isnull().sum()
```

```python
#vamos a imputar los datos para el campo Edad, calcula
X = X.copy()
X["Age"] = X["Age"].fillna(
    X["Age"].median()
)
```

```python
X["Embarked"] = X["Embarked"].fillna(
    X["Embarked"].mode()[0]
)
```

```python
X.isnull().sum()
```

## Parte 6. Normalizamos los datos categoricos (textos)

```python
X.head(2)
```

```python
X= pd.get_dummies(
    X,
    columns=[
        "Sex",
        "Embarked"
    ],
    drop_first=True
)
```

```python
X.head(2)
```

## Parte 7. Division de los datos

No debemos entrenar y evaluar utilizando los mismos datos.

80% para entrenar

20% para evaluar

```python
x_train, x_test, y_train, y_test = train_test_split(
X,
Y,
test_size=0.20,
random_state=42
)
```

```python
print("Entrenamiento: ", x_train.shape)
```

```python
print("Prueba X: ", x_test.shape)
```

```python
print("Entrenamiento Y: ", y_train.shape)
```

```python
print("Prueba Y: ", y_test.shape)
```

## ESCALADO DE LOS DATOS 
Las variables numericas tienen valores distintos. El escalado evita que las variables con mayores valores dominen el aprendizaje del modelos

```python
columnas_numericas = [
    "Age",
    "Fare"
]
```

```python
x_train[columnas_numericas].describe()
```

```python
#Escalamos los datos usando StandarScaler
scaler = StandardScaler()
```

```python
x_train[columnas_numericas]= scaler.fit_transform(x_train[columnas_numericas])
x_test[columnas_numericas]= scaler.transform(x_test[columnas_numericas])
```

```python
x_train[columnas_numericas].describe()
```

#Escalamos para que los datos no sean discriminados 

## Entrenamiento del algoritmo albol de descision
permite clasificar los registros con respectos a las variables

```python
modelo = DecisionTreeClassifier(
    max_depth=4,
    random_state=42
)
```

```python
modelo.fit(x_train, y_train)
```

## Evaluamos la eficacia del modelo

Realizamos predicciones sobre el conjunto de prueba, para calcular la exactitud del modelo

```python
prediccion = modelo.predict(
    x_test
)
```

```python
pd.DataFrame(
    {
        "Survived Real: " : y_test,
        "Survived Predicha: " : prediccion
    }
).head(20)
```

```python
accuracy = accuracy_score(y_test, prediccion)
```

```python
print(f"La exactitud del modelo es: {accuracy:.4f}")
```
