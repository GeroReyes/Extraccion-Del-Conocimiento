```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import (
    accuracy_score,
    confusion_matrix,
    classification_report,
    ConfusionMatrixDisplay
)
```

```python
iris = load_iris(as_frame=True)
```

```python
print(iris.DESCR)
```

```python
dfIris = iris.frame
dfIris.head(2)
```

```python
dfIris.tail(2)
```

```python
iris.target_names
```

```python
dfIris.rename(columns={"target" : "species"}, inplace=True)
```

```python
dfIris["species"] = dfIris["species"].map(
    {
        0: "setosa",
        1: "versicolor",
        2: "virginica"
    }
)
```

```python
dfIris.tail(5)
```

## Variables predictoras y variable objetivo 

X (Variables predictoras) son aquellas que me permiten hacer la prediccion 
Y (objetivo) Lo que quiero predecir

```python
x = dfIris.drop(columns=["species"])
```

```python
##variables objetivo
y = dfIris["species"]
```

```python
print(" ---- Variables predictoras ---- ")
print(x.columns)
print(" ---- Variable Objetivo ----- ")
print(y.name)
```

## Division del conjuntos de datos

*Training set:* Datos para entrenamiento.
*Testing set:* Datos no vistos por el modelo, para prueba

```python
X_train, X_test, y_train, y_test = train_test_split(
    x, 
    y,
    test_size = 0.20,
    random_state=42,
    stratify=y
)
```

```python
print("X_train: ", X_train.shape)
print("X_test: ", X_test.shape)
print("--------------------------")
print("y_train: ", y_train.shape)
print("y_test: ", y_test.shape)
```

```python
scaler = StandardScaler()
```

```python
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```python
print("-----Datos sin escalar-------- ")
print(X_train)
 
print("-------Escalados--------- ")   
print(X_train_scaled)
```

```python
print("-----------Datos sin escalar------------")
print(X_train.head(5))
print("-----------Escalados-----------")
print(X_train_scaled[:5])
```

```python
modelo = KNeighborsClassifier(
    n_neighbors = 3
)
```

```python
modelo.fit (
    X_train_scaled,
    y_train
)
```

```python
prediccion = modelo.predict(X_test_scaled)
```

```python
pd.DataFrame({
    "Especie real" : y_test.values,
    "Especie Predicha" : prediccion
}).head(20)
```
